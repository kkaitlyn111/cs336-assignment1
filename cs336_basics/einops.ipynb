```python
import numpy
import torch

# configure display settings for numpy arrays as images
def display_np_arrays_as_images():
    """configure display settings for numpy arrays as images"""
    try:
        import matplotlib.pyplot as plt
        plt.rcParams['figure.figsize'] = (10, 8)
        plt.rcParams['image.cmap'] = 'viridis'
        plt.rcParams['image.interpolation'] = 'nearest'
        print("matplotlib configured for displaying numpy arrays as images")
    except ImportError:
        print("matplotlib not available - numpy arrays will use default display settings")
        print("to install matplotlib: uv add matplotlib")

display_np_arrays_as_images()
```

```python
from einops import rearrange, reduce, repeat
```

```python
images = torch.randn(64, 128, 128, 3)
dim_by = torch.linspace(start=0.0, end=1.0, steps=10)
```

```python
dim_val = rearrange(dim_by, 'val -> 1 val 1 1 1')
imgs = rearrange(images, 'b h w c -> b 1 h w  c')
dimmed = imgs * dim_val
```
